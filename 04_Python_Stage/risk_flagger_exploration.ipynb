```python
import pandas as pd
import numpy as np

df_projects = pd.read_csv("../01_Raw_Data/campaigns_projects.csv")
df_logs = pd.read_csv("../01_Raw_Data/monthly_spend_kpis_dirty_for_python.csv")
```

```python
print(list(df_logs.columns))
```

```python
df_logs["project_id"] = df_logs["project_id"].astype(str).str.strip() 
```

```python
column_median = df_logs["amount_spent"].median()
df_logs["amount_spent"] = df_logs["amount_spent"].fillna(column_median)
```

```python
df_logs
```

```python
#Group df_logs by "Project_id" and to get total "amount_spent" and average "kpi_hit" 
```

```python
sum_per_project = df_logs.groupby("project_id")["amount_spent"].sum()
print(sum_per_project)
```

```python
avg_kpi_per_project = df_logs.groupby("project_id")["kpi_hit"].mean()
print(avg_kpi_per_project)
```

```python
print(sum_per_project, avg_kpi_per_project)
```

```python
#Merge the two dataframes to get a final dataframe with "project_id", "total_amount_spent" and "average_kpi_hit"
#Merge the smaller ones first 
df_avg = avg_kpi_per_project.reset_index()
df_sum = sum_per_project.reset_index()

merged_df = pd.merge(df_sum, df_avg , on  = "project_id")

print (merged_df)
```

```python
one_merged_df = merged_df.reset_index()
one_project_id = df_projects.reset_index()

final_df = pd.merge(one_merged_df, one_project_id , on  = "project_id")
print (final_df)
```

```python
df_projects.head(6)
```

```python
burn_rate_pct = final_df["amount_spent"] / final_df["budget_allocated"] * 100
print (burn_rate_pct)
final_df["burn_rate_pct"] = burn_rate_pct
```

```python
final_df["risk_flag"] = np.where((final_df["burn_rate_pct"] > 80) & (final_df["kpi_hit"] < 50), "critical_risk", "on_track")
print (final_df)
```

```python
final_df.to_excel("Automated_Executive_Risk_Report.xlsx",sheet_name="Risk_Report",index = False)
print("Done!")
```
